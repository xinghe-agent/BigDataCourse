# 实验一：静态网页爬取

## 案例一：QQ 表情包图片下载



```python
import requests
from pathlib import Path

# 1. 定义请求信息
url_list = [
    "http://www.bspider.top/static/yh31/images/202103302056530751.gif",
    "http://www.bspider.top/static/yh31/images/202103182029425871.jpg"
]

# 2. 发送请求并获得响应
def get_image(url):
    r = requests.get(url)
    return r.content

# 3. 将响应图片存储到本地
def save_image(path, content):
    with open(path, "wb") as f:
        f.write(content)

# 批量下载图片
def download_images(url_list):
    Path("qq_images").mkdir(exist_ok=True)
    for url in url_list:
        content = get_image(url)
        file_name = url.split("/")[-1]
        save_image(f"qq_images/{file_name}", content)

if __name__ == "__main__":
    download_images(url_list)
```

## 案例二：手机号码归属地查询

```python
import requests

# 1. 定义请求信息
phone = input("请输入自己的手机号码：")

url = "https://www.ip138.com/mobile.asp"
params = {
    "mobile": phone,
    "action": "mobile"
}
headers = {
    "User-Agent": "Mozilla/5.0"
}

# 2. 发送请求并获得响应
r = requests.get(url, params=params, headers=headers)
r.encoding = r.apparent_encoding
html = r.text

# 3. 将响应存储到本地 HTML
with open("phone_query_result.html", "w", encoding="utf-8") as f:
    f.write(html)
```
