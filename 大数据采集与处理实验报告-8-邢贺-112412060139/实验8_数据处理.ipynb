# 实验 8：数据处理

数据文件：`药品销售数据.csv`

使用 pandas 完成数据读取、缺失值处理、统计分析、衍生列计算和分组排序。

```python
import pandas as pd
```

## 要求 1：读取数据、删除缺失值并查看第一条记录

```python
df = pd.read_csv("药品销售数据.csv")
df = df.dropna()

print("清理后的数据：")
print(df.shape)
print("第一条记录：")
print(df.iloc[0])
```

## 要求 2：统计实收金额

计算实收金额的最大值、最小值、均值、总和和方差。

```python
amount = df["实收金额"]

print("最大值：", amount.max())
print("最小值：", amount.min())
print("均值：", amount.mean())
print("总和：", amount.sum())
print("方差：", amount.var())
```

## 要求 3：计算单价和折扣

单价 = 应收金额 / 销售数量；折扣 = 实收金额 / 应收金额。

```python
df["单价"] = df.apply(lambda row: row["应收金额"] / row["销售数量"], axis=1)
df["折扣"] = df.apply(lambda row: row["实收金额"] / row["应收金额"], axis=1)

print(df[["商品名称", "销售数量", "应收金额", "实收金额", "单价", "折扣"]].head())
```

## 要求 4：按商品名称统计总销量并降序排序

```python
sales = df.groupby("商品名称")["销售数量"].sum()
sales = sales.sort_values(ascending=False)

print(sales)
```

## 重要知识点

1. 读取 CSV 文件：`pd.read_csv()`。
2. 访问 DataFrame 的某一行：`iloc[]` 或 `loc[]`。
3. 计算最大值：`max()`。
4. 分组统计：`groupby()`。
5. 删除缺失值：`dropna()`。
6. 排序：`sort_values()`。
